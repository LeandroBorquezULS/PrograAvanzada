```python
import time
from pyspark.sql.types import (StructType, StructField, StringType,
                               DoubleType, LongType)

esquema_crudo = StructType([
    StructField("timestamp", StringType(), True),
    StructField("sensor_id", StringType(), True),
    StructField("metric",    StringType(), True),
    StructField("value",     DoubleType(), True),
    StructField("unit",      StringType(), True),
    StructField("worker_id", LongType(),   True),
    StructField("seq",       LongType(),   True),
])

t0 = time.perf_counter()
df_crudo = spark.read.schema(esquema_crudo).json("/opt/workspace/data/raw/*.jsonl")
total = df_crudo.count()
t1 = time.perf_counter()

print(f"✓ Registros leídos con esquema: {total:,} en {t1 - t0:.2f} s")
df_crudo.printSchema()
df_crudo.show(5)
```

```python
import time
from pyspark.sql import functions as F

# Desactivar AQE para observar el efecto de la configuración manual
spark.conf.set("spark.sql.adaptive.enabled", "false")

for num_parts in [200, 8]:
    spark.conf.set("spark.sql.shuffle.partitions", str(num_parts))
    t0 = time.perf_counter()
    resumen = df_crudo.groupBy("sensor_id").agg(F.avg("value").alias("promedio")).count()
    t_total = time.perf_counter() - t0
    print(f"Agrupación con shuffle ({num_parts} particiones): {t_total:.2f} segundos")

# Reactivar AQE recomendado
spark.conf.set("spark.sql.adaptive.enabled", "true")
```

```python
from pyspark.sql import functions as F

# 1. Ver el plan físico para detectar el Exchange (Shuffle)
df_crudo.groupBy("sensor_id").agg(F.avg("value").alias("promedio")).explain()

# 2. Comparación del impacto de particiones sin AQE
spark.conf.set("spark.sql.adaptive.enabled", "false")

for parts in [200, 8]:
    spark.conf.set("spark.sql.shuffle.partitions", str(parts))
    t_start = time.perf_counter()
    agg_df = df_crudo.groupBy("sensor_id").agg(F.avg("value")).count()
    duracion = time.perf_counter() - t_start
    print(f"Shuffle con {parts} particiones: {duracion:.2f} s")

# 3. Reactivar AQE
spark.conf.set("spark.sql.adaptive.enabled", "true")
print("✓ AQE reactivado")
```

```python
import os

if os.path.exists("/opt/workspace/salida_sitio3"):
    telemetria = spark.read.parquet("/opt/workspace/salida_sitio3/telemetria")
    etiquetas = spark.read.parquet("/opt/workspace/salida_sitio3/verdad_terreno")
    
    n_antes = telemetria.count()
    
    # Range Join para evitar multiplicar filas por fallas temporales
    cond = (
        (telemetria["entidad_id"] == etiquetas["entidad_id"]) &
        (telemetria["timestamp_utc"] >= etiquetas["t_inicio"]) &
        (telemetria["timestamp_utc"] <= etiquetas["t_fin"])
    )
    
    # Broadcast join con la tabla pequeña
    etiquetada = telemetria.join(F.broadcast(etiquetas), cond, "left")
    
    # Comprobación estricta de filas
    assert etiquetada.count() == n_antes, "ERROR: El join alteró el número de filas (fan-out)"
    print(f"✓ Join verificado sin duplicación de filas: {n_antes:,} registros.")
else:
    print("Directorio salida_sitio3 no encontrado. Omitiendo join.")
```

```python
print("=== PERFILADO DE DATA/RAW/ (ÍTEM 8) ===")

# 1. Comprobación de nulos por columna
nulos = [df_crudo.filter(F.col(c).isNull()).count() for c in df_crudo.columns]
print(f"Nulos por columna: {dict(zip(df_crudo.columns, nulos))}")

# 2. Conteo de métricas y unidades distintas
df_crudo.select("metric", "unit").distinct().show()

# 3. Rango de valores
df_crudo.select(
    F.min("value").alias("min_valor"),
    F.max("value").alias("max_valor"),
    F.avg("value").alias("promedio_valor")
).show()

# 4. Distribución por entidad/sensor (detección de Data Skew)
print("Distribución de eventos por sensor:")
df_crudo.groupBy("sensor_id").count().orderBy(F.desc("count")).show(20)
```
